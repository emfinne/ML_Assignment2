```python
import pandas as pd

def load_dataset():
    return pd.read_csv(
        "diabetes_binary_classification_data.csv"
    )

df = load_dataset()
```

```python
def plot_binary_pie1(df, feature):
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))

    for value, ax in zip([0, 1], axes):
        subset = df[df[feature] == value]
        counts = subset["Diabetes_binary"].value_counts().reindex(
            [0, 1], fill_value=0
        )

        ax.pie(
            counts,
            labels=["No diabetes", "Diabetes"],
            autopct="%1.1f%%",
            colors=["steelblue", "coral"],
            startangle=90
        )
        ax.set_title(f"{feature} = {value} (n={len(subset):,})")

    plt.tight_layout()
    plt.show()
```

```python
def plot_ordinal(df, feature, labels=None):
    percentages = pd.crosstab(
        df[feature], df["Diabetes_binary"],
        normalize="columns"
    ).reindex(columns=[0, 1], fill_value=0) * 100

    percentages.columns = ["No diabetes", "Diabetes"]

    if labels:
        percentages.index = [
            labels.get(value, value)
            for value in percentages.index
        ]

    ax = percentages.plot.bar(
        figsize=(10, 5),
        color=["steelblue", "coral"],
        rot=0
    )

    ax.set_xlabel(feature)
    ax.set_ylabel("Percentage within each diabetes group")
    ax.set_title(f"Distribution of {feature} by diabetes status")
    ax.set_ylim(0, 100)

    for bars in ax.containers:
        ax.bar_label(bars, fmt="%.1f%%", padding=3)

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
def plot_numerical(df, feature, bins=20):
    bin_edges = np.histogram_bin_edges(df[feature].dropna(), bins=bins)
    fig, ax = plt.subplots(figsize=(10, 5))

    for value, label, color, layer in [
        (0, "No diabetes", "steelblue", 1),
        (1, "Diabetes", "coral", 2)
    ]:
        values = df.loc[
            df["Diabetes_binary"] == value, feature
        ].dropna()

        ax.hist(
            values,
            bins=bin_edges,
            weights=np.full(len(values), 100 / len(values)),
            color=color,
            alpha=0.5,
            edgecolor=color,
            label=label,
            zorder=layer
        )

    ax.set_xlabel(feature)
    ax.set_ylabel("Respondents within each group (%)")
    ax.set_title(f"Distribution of {feature} by diabetes status")
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt

def plot_hist(df, feature, bins=40, density=True):

    df[df["Diabetes_binary"] == 0][feature].hist(
        bins=bins, alpha=0.5, density=density, label="No diabetes"
    )

    df[df["Diabetes_binary"] == 1][feature].hist(
        bins=bins, alpha=0.5, density=density, label="Diabetes"
    )

    plt.xlabel(feature)
    plt.ylabel("Density" if density else "Count")
    plt.title(f"Distribution of {feature}")
    plt.legend()
    plt.show()
```
