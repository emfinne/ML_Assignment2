Functions and other code related stuff below:

```python
import pandas as pd

def load_dataset():
    return pd.read_csv(
        "diabetes_binary_classification_data.csv"
    )

df = load_dataset()
```

```python
import matplotlib.pyplot as plt

def plot_hist(df, feature, bins=40, density=True):

    df[df["Diabetes_binary"] == 0][feature].hist(
        bins=bins, alpha=0.5, density=density, label="No diabetes"
    )

    df[df["Diabetes_binary"] == 1][feature].hist(
        bins=bins, alpha=0.5, density=density, label="Diabetes"
    )

    plt.xlabel(feature)
    plt.ylabel("Density" if density else "Count")
    plt.title(f"Distribution of {feature}")
    plt.legend()
    plt.show()
```

```python
import matplotlib.pyplot as plt

def plot_binary_pie(df, feature, target="Diabetes_binary", labels=None):

    fig, ax = plt.subplots(1, 2, figsize=(12, 5))

    for i in [0, 1]:

        # Select observations based on diabetes status
        subset = df[df[target] == i]

        # Count the feature categories
        counts = subset[feature].value_counts().reindex([0, 1], fill_value=0)

        # Create descriptive labels
        category_labels = [
            labels.get(j, str(j)) if labels else f"{feature} = {j}"
            for j in [0, 1]
        ]

        ax[i].pie(
            counts,
            labels=category_labels,
            autopct="%1.1f%%",
            startangle=90,
            colors=["steelblue", "coral"]
        )

        title = "No diabetes" if i == 0 else "Diabetes"
        ax[i].set_title(f"{title}\n(n={len(subset):,})")

    fig.suptitle(f"Distribution of {feature} by diabetes status")

    plt.tight_layout()
    plt.show()
```
