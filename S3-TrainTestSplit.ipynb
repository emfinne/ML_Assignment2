# Creating the Train/Test split

Importing the data after analysis.

The whole dataset is already fully encoded. As such, there is nothing for us to encode.

```python
import pandas as pd
import matplotlib.pyplot as plt

def load_dataset():
    return pd.read_csv(
        "diabetes_binary_classification_data.csv" # Replace this with the preprocessed
    )

df = load_dataset()
```

Then we split the dataset, while keeping the distributions of `Diabetes_binary` the same between those two.
This is performed by stratifying with respect to `Diabetes_binary`.

```python
from sklearn.model_selection import train_test_split

# Stratify diabetes
sugar_overdose_group = pd.qcut(
    df["Diabetes_binary"],
    q=2,
    labels=False
)

train_set, test_set = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=sugar_overdose_group
)

print(train_set.shape, test_set.shape)

print("\nTraining Diabetes_binary distribution:")
print(train_set["Diabetes_binary"].describe())

print("\nTest Diabetes_binary distribution:")
print(test_set["Diabetes_binary"].describe())
```

Saving the files for the separate datasets. 

```python
train_set.to_csv("train_set.csv", index=False)
test_set.to_csv("test_set.csv", index=False)
```
