# Creating the Train/Test split

Importing the data after analysis.

The whole dataset is already fully encoded. As such, there is nothing for us to encode.

```python
import pandas as pd
import matplotlib.pyplot as plt

def load_dataset():
    return pd.read_csv(
        "diabetes_binary_classification_data.csv" # Replace this with the preprocessed
    )

df = load_dataset()
```

Then we split the dataset, while keeping the distributions of `Diabetes_binary` the same between those two.
This is performed by stratifying with respect to `Diabetes_binary`.

```python
from sklearn.model_selection import train_test_split

#The "Diabetes_binary" feature is already grouped by every row's value.
#We can insert it directly into the stratify parameter
train_set, test_set = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["Diabetes_binary"]
)

train_set_description = train_set["Diabetes_binary"].describe()
test_set_description = test_set["Diabetes_binary"].describe()

print("Train shape: ", train_set.shape, " Test shape: ", test_set.shape)

print("\nMean and standard deviation:")
print("\tTrain \tTest")
means = (train_set_description.loc["mean"], test_set_description.loc["mean"])
stds = (train_set_description.loc["std"], test_set_description.loc["std"])
print(f"mean:\t{means[0]:.5f}\t{means[1]:.5f}")
print(f"std:\t{stds[0]:.5f}\t{stds[1]:.5f}")

#print("\nTraining Diabetes_binary distribution:")
#print(train_set_description)

#print("\nTest Diabetes_binary distribution:")
#print(test_set_description)
```

As we can see, the mean and standard deviation of both sets is pretty much the same, which means the stratification succeeded.
If the dataset was smaller, the numbers would likely end up much more different from each other.

Then, we save the two split datasets for other the next steps to use now that we are done here.

```python
train_set.to_csv("generated/train_set.csv", index=False)
test_set.to_csv("generated/test_set.csv", index=False)
```
