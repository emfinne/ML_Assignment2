# Evaluating the Model



After using the model, we must asses its performance.

```python
import matplotlib.pyplot as plt
from sklearn import metrics

yTrue = [0, 0, 1, 1, 0, 1, 1, 0] #True answers, as per the train dataset
yPredicted = [0, 1, 0, 1, 1, 1, 0, 1] #Answers predicted by the model
```

## Confusion Matrix

```python
cm = metrics.confusion_matrix(yTrue, yPredicted)

TP = cm[0, 0]
FN = cm[1, 0]
TN = cm[1, 1]
FP = cm[0, 1]

disp = metrics.ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.show()
```

## Accuracy

How many answers did the model get right?

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)
```

## Precision

How many did the model flag who actually have diabetes?
Aka. reliability of disease alerts.

```python
precision = TP / (TP + FP)
```

## Recall

How many who have diabetes were correctly identified by the model?
Aka. how many disease cases the model thinks it found.

```python
recall = TP / (TP + FN)
```

## F1 Score

Summarizes *precision* and *recall* into one number.
A value of *100%* means perfect *precision* and *recall*. 

```python
f1_score = 2 * (precision * recall) / (precision + recall)
```
